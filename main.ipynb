```python
from data_processor import document_from_text, extract_document
from inference import analyze_document
document = document_from_text("가상기업 (123456)\n발간일: 2025.01.15\n목표주가: 82,000원\n\n수요 둔화로 실적 전망을 하향한다.")
result = analyze_document(document, engine="rules")
result["metrics"]
```
